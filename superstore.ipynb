```python
df.columns=df.columns.str.strip()
df.columns=df.columns.str.replace(" ","_")
df.columns=df.columns.str.replace("-","_")
df.columns=df.columns.str.capitalize()
```

```python
df.info()
```

```python
df["Order_date"]=pd.to_datetime(df["Order_date"],format="mixed")
df["Ship_date"]=pd.to_datetime(df["Ship_date"],format="mixed")
```

```python
df.drop(["Row_id","Product_id","Postal_code"], axis=1, inplace=True)
```

```python
df["Month"]=df["Order_date"].dt.month
df["Year"]=df["Order_date"].dt.year
```

```python
df.groupby("Region")["Profit"].agg(["sum","mean"]).reset_index()
```

```python
df.groupby("Region")["Profit"].sum().plot(kind="bar")
plt.xlabel("Religion")
plt.ylabel("Profit")
plt.title("Relion wise Profit")
plt.show()
```

```python
df.groupby("Region")["Sales"].agg(["sum","mean"]).reset_index()
```

```python
df.groupby("Region")["Sales"].sum().plot(kind="bar")
plt.xlabel("Religion")
plt.ylabel("Sales")
plt.title("Relion wise Sales")
plt.show()
```

```python
df.groupby("Month")[["Sales","Profit"]].sum().reset_index()
```

```python
plt.figure(figsize=(10,6))

plt.plot(df.groupby("Month")["Profit"].sum().index,df.groupby("Month")["Profit"].sum().values,marker="*")

plt.xlabel("Month")
plt.ylabel("Profit")
plt.title("Monthly Profit")
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.plot(df.groupby("Month")["Sales"].sum().index,df.groupby("Month")["Sales"].sum().values,marker="*")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()
```
